#### Problem 1: Zipper file()

Given a string filename and a list of strings, create a file that has on each line a zippered string consisting of sequential pairs of strings. <br>
So the first line would have strings at index 0 and 1 zippered together, line 2 would have strings 2 and 3, and so on... <br>
You may assume that the input list is always an even length(so every string has a pair).

```python
def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            s1 = strings[i]
            s2 = strings[i + 1]

            zipped = ""

            for j in range(max(len(s1), len(s2))):
                if j < len(s1):
                    zipped += s1[j]

                if j < len(s2):
                    zipped += s2[j]

            file.write(zipped + "\n")

strings = ["ABC", "123", "DEF", "456"]

zipper_file("output.txt", strings)

print("File created successfully!\n")

print("File Content: \n")
with open("output.txt", "r") as file:
    print(file.read())
```

#### Problem 2: Reformat date() 

Brad's back with more class infrastructure problems that he needs you to solve!<br>

In this case Brad has been sitting on a massive list of dates corresponding to office hour visits but he's completely forgotten how to read the datestring format "MM/DD/YYYY"! <br>
Remind Brad about what month corresponds to which number (from 00 to 12) by taking as input a list of strings (MM/DD/YYYY format) and <br> 
converting them into English sentences of the form "Office Hours on Month DD, YYYY". <br> 

    Note: You should not test for the validity of dates. Brad[the autograder] lives in a world where every month has 31 days. :)
          Make sure your output precisely matches the required format, such as in the following examples.
          Notably, the days should remain a two digit string!

```python
def reformat_date(dates):

    months = ["January", "February", "March", "April", "May", "June", "July",
              "August", "September", "October", "November", "December"]

    result = []

    for date in dates:
        month, day, year = date.split("/")

        month_name = months[int(month) - 1]

        formatted_date = "Office Hours on " + month_name + " " + day + ", " + year

        result.append(formatted_date)

    return result

dates = ["01/15/2026", "03/09/2026", "12/25/2026"]

result = reformat_date(dates)

for date in result:
    print(date)
```
