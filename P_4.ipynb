4. ` Write a Python program in a Jupyter Notebook that takes an integer input (1 to 9999) from
the user and determines the following using branching statements:` <br>
    1. If the number is even or odd.
    2. If the number is a palindrome (reads the same forward and backward).
    3. If the number is divisible by the Sum Of its digits.

```python
num = int(input("Enter an integer between 1 and 9999: "))

# Check even or odd

if(num % 2 == 0):
    print("The number is even.")
else:
    print("The number is odd.")

# Check Palindrome

if str(num) == str(num)[::-1]:
    print("The number is a palindrome.")
else:
    print("The number is not a palindrome.")

# Sum of digits
digit_sum = sum(int(digit) for digit in str(num))

# Divisible by sum of digits

if num % digit_sum == 0:
    print("The number is divisible by the sum of its digits.")
else:
    print("The number is not divisible by the sum of its digits.")


```

7. ` Write a while loop to copy the strings 'orange' of the list to the list squares to the list new_squares. Stop and exit the loop if the value on the list is not 'orange'`

```python
squares = ['red', 'yellow', 'green', 'purple', 'blue']

new_squares = []

i = 0

while i < len(squares):
    if squares[i] != 'orange':
        break

    new_squares.append(squares[i])
    i += 1

print(new_squares)
```

8. `The following is a list of animals in a National Zoo: `<br>
     `Animals = ["lion", "giraffe", "gorilla", "parrots", "crocodile", "deer", "swan"]` <br>
Your brother needs to write an essay on the animals whose names are made of 7 letters. Help him find those animals through a while loop and create a separate list of such animals.

```python
animals = ["lion", "giraffe", "gorilla", "parrots",
           "crocodile", "deer", "swan"]

seven_letter_animals = []

i = 0

while i < len(animals):
    if len(animals[i]) == 7:
        seven_letter_animals.append(animals[i])

    i += 1

print(seven_letter_animals)
```
