`Fixing the above problem with Super in Python`    

```python
class person:

    # Constructor
    def __init__(self, name, id):
        self.name = name 
        self.id = id 

    # To check if this person is an employee 
    def Display(self):
        print(self.name, self.id)


class Emp(person):

    def __init__(self, name, id):
        super().__init__(name, id)
        self.name_ = name

    def Print(self):
        print("Emp class called")


Emp_details = Emp("Mayank", 103)

# Calling parent class function
Emp_details.Display()

print(Emp_details.name_)
print(Emp_details.name)
```

#### `Problem 2: Write a Python program to find Anagram Check`

What is an anagram?

Two words/strings are anagrams if they contain:

* the same characters
* with the same frequency/number of occurrences
* but the characters can be in a different order

Example : 

listen
silent

Compare their letters:

listen → l i s t e n

silent → s i l e n t

Both exactly contains:

l, i, s, t, e, n

So:

listen and silent → Anagrams

```python
str1 = input("Enter the first string: ")
str2 = input("Enter the second string: ")

str1 = str1.replace(" ", "").lower()
str2 = str2.replace(" ", "").lower()

if sorted(str1) == sorted(str2):
    print("Anagrams")
else:
    print("Not Anagrams")
```

#### `Problem 3: Write a Python program to check for Armstrong Number.`

Armstrong Number:
    
A number that is equal to the sum of its own digits, each raised to the power of the total number of digits.

* 3-Digit Example (153):
  
    * Total digits (n) = 3
    * Calculation: (1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153
    * Result: 153 is an Armstrong number.

```python
num = int(input("Enter the number: "))

original = num

digits = len(str(num))
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit ** digits
    num = num // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

#### `Problem 4: Deal with Classes and Objects` 

Scenario: Car dealership's inventory management system 

Task-1. You are tasked with creating a Python program to represent vehicles using a class. Each car should have attributes for maximum speed and mileage.

Task-2. Update the class with the default color for all vehicles, "white".

Task-3. Additionally, you need to create methods in the vehicle class to assign seating capacity to a vehicle.

Task-4. Create a method to display all the properties of an object of the class.

Task-5. Additionally, you need to create two objects of the vehicle class object that should have a max speed of 200kmph and mileage of 20kmpl with five seating capacities, and another car object should have a max speed of 180kmmph and mileage of 25kmpl with four seating capacities. 

```python
class Vehicle:

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage 
        self.color = "white"

    def seating_capacity(self, capacity):
        self.seating_capacity = capacity


    def display(self):
        print("Maximum Speed: ", self.max_speed)
        print("Mileage: ", self.mileage)
        print("Color: ", self.color)
        print("Seating Capacity: ", self.seating_capacity)



car1 = Vehicle(200, 20)
car1.seating_capacity(5)

car2 = Vehicle(180, 25)
car2.seating_capacity(4)

print("Car 1: ")
car1.display()

print("\nCar 2: ")
car2.display()
```
